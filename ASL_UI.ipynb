```python
from ui.components import SignSightDemo

app = SignSightDemo()
app.show()
```

```python
sample_prediction = {
    "hand_visible": True,
    "label": "A",
    "confidence": 0.97,
    "top3": [("A", 0.97), ("S", 0.02), ("E", 0.01)],
}

app.start()
app.prediction = sample_prediction
app.logic.process(sample_prediction)
app.render()

print("Sent one simulated A prediction to SignSight.")
```
